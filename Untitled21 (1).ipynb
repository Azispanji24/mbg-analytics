## Tahap 4: Data Cleaning / Quality Check
Mengecek apakah ada file gambar duplikat (berdasarkan hash konten file)
untuk menghindari transaksi ganda saat analisis asosiasi.

```python
import hashlib

def file_hash(filepath):
    with open(filepath, 'rb') as f:
        return hashlib.md5(f.read()).hexdigest()

hash_map = {}
duplicates = []

for fname in image_files:
    h = file_hash(os.path.join(FOLDER_PATH, fname))
    if h in hash_map:
        duplicates.append((fname, hash_map[h]))
    else:
        hash_map[h] = fname

print(f"Jumlah duplikat ditemukan: {len(duplicates)}")
print(duplicates[:10])
```

```python
duplicate_files = [d[0] for d in duplicates]  # ambil nama file yang mau dibuang

image_files = [f for f in image_files if f not in duplicate_files]
print(f"Total gambar setelah hapus duplikat: {len(image_files)}")
```

## Tahap 5: Preprocessing Gambar
Resize gambar ke ukuran standar untuk mempercepat proses OCR,
tanpa mengurangi keterbacaan teks.

```python
PROCESSED_PATH = '/content/dataset_resized'
os.makedirs(PROCESSED_PATH, exist_ok=True)

MAX_WIDTH = 1000  # cukup besar buat OCR tetap akurat, tapi gak terlalu berat

for fname in image_files:
    img_path = os.path.join(FOLDER_PATH, fname)
    img = Image.open(img_path).convert('RGB')

    if img.width > MAX_WIDTH:
        ratio = MAX_WIDTH / img.width
        new_size = (MAX_WIDTH, int(img.height * ratio))
        img = img.resize(new_size)

    save_path = os.path.join(PROCESSED_PATH, fname.rsplit('.', 1)[0] + '.png')
    img.save(save_path)

print("Selesai resize. Total file:", len(os.listdir(PROCESSED_PATH)))
```

## Tahap 7: Cleaning Hasil OCR & Ekstraksi Item Menu
Karena hasil OCR mengandung noise (info gizi, nama SPPG, tanggal, dll),
dilakukan pencocokan kata terhadap kamus item makanan (whitelist) untuk
mengekstrak item menu yang valid dari tiap gambar.

```python
# Kamus item makanan — bisa terus ditambah sambil ngecek hasil OCR lebih banyak sample
FOOD_KEYWORDS = {
    'nasi': ['nasi'],
    'telur': ['telur'],
    'roti': ['roti'],
    'susu': ['susu', 'milk', 'uht'],
    'pisang': ['pisang', 'pisong'],
    'jeruk': ['jeruk'],
    'ubi': ['ubi'],
    'kurma': ['kurma'],
    'kacang': ['kacang'],
    'tempe': ['tempe'],
    'abon': ['abon'],
    'sayur': ['sayur', 'sawi', 'buncis', 'tumis'],
    'jamur': ['jamur'],
    'rendang': ['rendang'],
    'ayam': ['ayam'],
    'ikan': ['ikan'],
    'tahu': ['tahu'],
    'kentang': ['kentang'],
    'apel': ['apel'],
    'semangka': ['semangka'],
    'melon': ['melon'],
    'pepaya': ['pepaya'],
    'sup': ['sup', 'soto'],
    'mie': ['mie', 'bihun'],
}

def extract_items(ocr_words):
    text = ' '.join(ocr_words).lower()
    found = set()
    for item, keywords in FOOD_KEYWORDS.items():
        for kw in keywords:
            if kw in text:
                found.add(item)
                break
    return list(found)

# Test pakai hasil OCR sample tadi
sample_ocr = ['BADAN', 'GIZI', 'SABTU', 'NASIONAL', '07 FEBRUARI 2026', 'Roti isi kacang hijau', 'Susu', 'Telur ayam rebus', 'Pisong']
print(extract_items(sample_ocr))
```

```python
import json
import time

ocr_results = {}
all_files = sorted(os.listdir(PROCESSED_PATH))

start = time.time()
for i, fname in enumerate(all_files):
    path = os.path.join(PROCESSED_PATH, fname)
    try:
        result = reader.readtext(path, detail=0)
        ocr_results[fname] = result
    except Exception as e:
        ocr_results[fname] = []
        print(f"Gagal OCR {fname}: {e}")

    if (i+1) % 50 == 0:
        elapsed = time.time() - start
        print(f"Progress: {i+1}/{len(all_files)} | Elapsed: {elapsed:.1f}s")

print("Selesai OCR semua gambar.")

# Simpan hasil mentah ke JSON biar gak perlu OCR ulang kalau Colab disconnect
with open('/content/drive/MyDrive/ML/ocr_results_raw.json', 'w', encoding='utf-8') as f:
    json.dump(ocr_results, f, ensure_ascii=False, indent=2)

print("Hasil OCR disimpan ke Drive: ocr_results_raw.json")
```

```python
import re
from collections import Counter

word_counter = Counter()

for fname, words in ocr_results.items():
    for w in words:
        # Bersihkan: lowercase, hilangkan angka & simbol, split kalau ada multi-kata
        clean = re.sub(r'[^a-zA-Z\s]', ' ', w.lower())
        tokens = clean.split()
        for t in tokens:
            if len(t) > 2:  # buang token terlalu pendek (noise)
                word_counter[t] += 1

print("Total kata unik:", len(word_counter))
print("\nTop 100 kata paling sering muncul:")
for word, count in word_counter.most_common(100):
    print(f"{word}: {count}")
```

```python
# Load hasil OCR dari JSON (kalau Colab reconnect, jalankan ini dulu)
import json

with open('/content/drive/MyDrive/ML/ocr_results_raw.json', 'r', encoding='utf-8') as f:
    ocr_results = json.load(f)

print(f"Total gambar di OCR results: {len(ocr_results)}")
```

```python
# Terapkan extract_items ke semua hasil OCR
transactions_raw = {}

for fname, ocr_words in ocr_results.items():
    items = extract_items(ocr_words)
    transactions_raw[fname] = items

# Cek hasilnya
total = len(transactions_raw)
kosong = sum(1 for v in transactions_raw.values() if len(v) == 0)
print(f"Total transaksi: {total}")
print(f"Transaksi kosong (0 item terdeteksi): {kosong}")
print(f"Transaksi valid: {total - kosong}")
print()
print("Contoh 5 transaksi pertama:")
for fname, items in list(transactions_raw.items())[:5]:
    print(f"  {fname}: {items}")
```

```python
# Buang transaksi kosong
transactions = {fname: items for fname, items in transactions_raw.items() if len(items) > 0}

print(f"Total transaksi setelah filter: {len(transactions)}")

# Simpan ke Drive biar aman
with open('/content/drive/MyDrive/ML/transactions_clean.json', 'w', encoding='utf-8') as f:
    json.dump(transactions, f, ensure_ascii=False, indent=2)

print("Transaksi bersih disimpan ke: transactions_clean.json")
```

## Tahap 8: Transformasi ke Data Transaksi
Setiap gambar dikonversi menjadi satu transaksi berisi list item makanan
yang sudah bersih. Format ini adalah input standar untuk algoritma
Association Rule Mining.

```python
# Ubah dict ke list of lists (format yang dibutuhkan TransactionEncoder)
transaction_list = list(transactions.values())

print(f"Total transaksi: {len(transaction_list)}")
print(f"\nContoh 5 transaksi:")
for t in transaction_list[:5]:
    print(f"  {t}")

# Cek distribusi jumlah item per transaksi
from collections import Counter
item_counts = Counter(len(t) for t in transaction_list)
print(f"\nDistribusi jumlah item per transaksi:")
for k in sorted(item_counts):
    print(f"  {k} item: {item_counts[k]} transaksi")
```

## Tahap 9: Encoding Data (One-Hot Encoding Transaksi)
Data transaksi dikonversi ke format one-hot encoding menggunakan
TransactionEncoder dari mlxtend, sehingga setiap item menjadi kolom
dengan nilai True/False per transaksi.

```python
from mlxtend.preprocessing import TransactionEncoder
import pandas as pd

te = TransactionEncoder()
te_array = te.fit(transaction_list).transform(transaction_list)
df = pd.DataFrame(te_array, columns=te.columns_)

print(f"Shape dataframe: {df.shape}")
print(f"Item unik: {list(df.columns)}")
print(f"\nPreview dataframe:")
df.head()
```

## Tahap 10: Penerapan Algoritma Asosiasi (FP-Growth)
Algoritma FP-Growth digunakan untuk menemukan frequent itemsets,
yaitu kombinasi item makanan yang sering muncul bersamaan dalam
menu MBG. Minimum support 0.05 artinya kombinasi harus muncul
di minimal 5% dari total transaksi (~42 dari 830 transaksi).

```python
from mlxtend.frequent_patterns import fpgrowth

frequent_itemsets = fpgrowth(df, min_support=0.05, use_colnames=True)
frequent_itemsets['length'] = frequent_itemsets['itemsets'].apply(len)
frequent_itemsets = frequent_itemsets.sort_values('support', ascending=False).reset_index(drop=True)

print(f"Total frequent itemsets ditemukan: {len(frequent_itemsets)}")
print(f"\nDistribusi panjang itemset:")
print(frequent_itemsets['length'].value_counts().sort_index())
print(f"\nTop 20 frequent itemsets:")
print(frequent_itemsets.head(20).to_string())
```

## Tahap 11: Generate Association Rules
Dari frequent itemsets yang ditemukan, dibangkitkan association rules
menggunakan metrik confidence dan lift. Minimum confidence 0.6 artinya
jika item A ada, maka item B muncul di minimal 60% kasus.
Lift > 1 menunjukkan hubungan positif antar item.

```python
from mlxtend.frequent_patterns import association_rules

rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.6)
rules = rules.sort_values('lift', ascending=False).reset_index(drop=True)

print(f"Total rules ditemukan: {len(rules)}")
print(f"\nKolom yang tersedia: {list(rules.columns)}")
print(f"\nTop 20 rules (berdasarkan lift tertinggi):")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(20).to_string())
```

## Tahap 12: Filtering & Sorting Rules
Menyaring rules yang paling signifikan berdasarkan kombinasi
lift tinggi (hubungan kuat antar item) dan confidence tinggi
(tingkat kepastian rule). Rules dengan lift > 1.5 dan
confidence > 0.65 dianggap paling bermakna.

```python
# Filter rules terbaik: lift > 1.5 dan confidence > 0.65
rules_filtered = rules[(rules['lift'] > 1.5) & (rules['confidence'] > 0.65)].copy()
rules_filtered = rules_filtered.sort_values('lift', ascending=False).reset_index(drop=True)

print(f"Total rules setelah filter: {len(rules_filtered)}")
print(f"\nRules terbaik (lift > 1.5 & confidence > 0.65):")
print(rules_filtered[['antecedents', 'consequents', 'support', 'confidence', 'lift']].to_string())

# Simpan ke CSV
rules_filtered.to_csv('/content/drive/MyDrive/ML/association_rules.csv', index=False)
print("\nRules disimpan ke: association_rules.csv")
```

## Tahap 13: Visualisasi Hasil
Visualisasi dilakukan dalam dua bentuk:
1. Bar chart frekuensi item makanan yang paling sering muncul
2. Scatter plot hubungan support vs confidence vs lift dari semua rules
3. Network graph untuk menampilkan hubungan antar item makanan

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

# ── 1. Bar chart frekuensi item ──────────────────────────────
item_support = frequent_itemsets[frequent_itemsets['length'] == 1].copy()
item_support['item'] = item_support['itemsets'].apply(lambda x: list(x)[0])
item_support = item_support.sort_values('support', ascending=True)

fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.barh(item_support['item'], item_support['support'] * 100,
               color='steelblue', edgecolor='white')
ax.set_xlabel('Support (%)', fontsize=12)
ax.set_title('Frekuensi Item Makanan dalam Menu MBG', fontsize=14, fontweight='bold')
for bar, val in zip(bars, item_support['support'] * 100):
    ax.text(val + 0.5, bar.get_y() + bar.get_height()/2,
            f'{val:.1f}%', va='center', fontsize=9)
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/ML/viz_item_frequency.png', dpi=150)
plt.show()

# ── 2. Scatter plot support vs confidence vs lift ────────────
fig, ax = plt.subplots(figsize=(10, 6))
sc = ax.scatter(rules['support'] * 100, rules['confidence'] * 100,
                c=rules['lift'], cmap='YlOrRd', s=100, edgecolors='gray', linewidth=0.5)
plt.colorbar(sc, ax=ax, label='Lift')
ax.set_xlabel('Support (%)', fontsize=12)
ax.set_ylabel('Confidence (%)', fontsize=12)
ax.set_title('Support vs Confidence (warna = Lift)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/ML/viz_support_confidence.png', dpi=150)
plt.show()

# ── 3. Network graph rules terbaik ───────────────────────────
import networkx as nx

G = nx.DiGraph()
for _, row in rules_filtered.iterrows():
    ant = ', '.join(list(row['antecedents']))
    con = ', '.join(list(row['consequents']))
    G.add_edge(ant, con, weight=round(row['lift'], 2),
               label=f"conf={row['confidence']:.2f}\nlift={row['lift']:.2f}")

fig, ax = plt.subplots(figsize=(12, 7))
pos = nx.spring_layout(G, seed=42, k=2)
nx.draw_networkx_nodes(G, pos, node_size=2500, node_color='steelblue', alpha=0.9, ax=ax)
nx.draw_networkx_labels(G, pos, font_size=10, font_color='white', font_weight='bold', ax=ax)
nx.draw_networkx_edges(G, pos, edge_color='gray', arrows=True,
                       arrowsize=20, width=2, ax=ax,
                       connectionstyle='arc3,rad=0.1')
edge_labels = nx.get_edge_attributes(G, 'label')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=8, ax=ax)
ax.set_title('Network Graph Association Rules Menu MBG', fontsize=14, fontweight='bold')
ax.axis('off')
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/ML/viz_network_graph.png', dpi=150)
plt.show()

print("Semua visualisasi disimpan ke Drive.")
```
